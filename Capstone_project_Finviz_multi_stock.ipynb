```python
news_data = []

for stock in stocks:

    # Extract Values
    industry = stock[0]

    ticker = stock[1]

    company_name = stock[2]

    print("Pulling Data For:", ticker)

    # Delay
    time.sleep(2)

    # Dynamic URL
    if ticker in ['BTCUSD', 'ETHUSD']:

        url = f"https://finviz.com/crypto_charts.ashx?t={ticker}&p=d"

    else:

        url = f"https://finviz.com/quote.ashx?t={ticker}"

    print(url)

    # Headers
    headers = {

        "User-Agent": (
            "Mozilla/5.0 "
            "(Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 "
            "(KHTML, like Gecko) "
            "Chrome/120.0.0.0 Safari/537.36"
        )
    }

    # Request
    response = requests.get(

        url,

        headers=headers
    )

    print("Status Code:", response.status_code)

    # Skip Failed Requests
    if response.status_code != 200:

        print("Failed:", ticker)

        continue

    # BeautifulSoup
    soup = BeautifulSoup(

        response.text,

        'html.parser'
    )

    # Find News Table
    news_table = soup.find(

        'table',

        id='news-table'
    )

    # Skip If No Table
    if news_table is None:

        print("No News Found")

        continue

    # Extract Rows
    rows = news_table.findAll('tr')

    print("Rows Found:", len(rows))

    # Loop Through Rows
    for row in rows:

        try:

            # Headline
            headline = row.a.text.strip().lower()

            # URL
            news_url = row.a.get('href')

            # Source
            source = row.span.text.strip().lower()

            source = re.sub(

                r"[()]",

                "",

                source
            )

            # Date & Time
            date_data = row.td.text.strip().split()

            if len(date_data) == 1:

                news_date = "Today"

                news_time = date_data[0]

            else:

                news_date = date_data[0]

                news_time = date_data[1]

            # Convert Time
            try:

                news_time = datetime.strptime(

                    news_time,

                    "%I:%M%p"

                ).strftime("%H:%M:%S")

            except:

                pass

            # Store Data
            news_data.append([

                industry,

                ticker,

                company_name,

                news_date,

                news_time,

                headline,

                source,

                news_url

            ])

        except Exception as e:

            print("Error:", e)

            continue

# Create DataFrame
columns = [

    'industry',
    'ticker',
    'company_name',
    'date',
    'time',
    'headline',
    'source',
    'url'

]

df = pd.DataFrame(

    news_data,

    columns=columns
)

# Check Output
print(df.shape)

df.head()
```

```python
df = df.drop_duplicates(

    subset=['ticker', 'headline']
)

df = df.reset_index(drop=True)
```

```python
print(df.shape)

df.head()
```

```python
today_date = datetime.today().strftime('%b-%d-%y')

df['date'] = df['date'].replace(

    'Today',

    today_date
)
```

```python
df['date'] = pd.to_datetime(

    df['date'],

    format='%b-%d-%y',

    errors='coerce'
)
```

```python
df['date'] = df['date'].ffill()
```

```python
df[['ticker', 'date']].head(10)
```

```python
df['time'] = pd.to_datetime(

    df['time'],

    format='%H:%M:%S',

    errors='coerce'
)
```

```python
df['year'] = df['date'].dt.year

df['month'] = df['date'].dt.month

df['month_name'] = df['date'].dt.month_name()

df['quarter'] = df['date'].dt.quarter

df['day'] = df['date'].dt.day

df['weekday'] = df['date'].dt.day_name()
```

```python
df['hour'] = df['time'].dt.hour

df['minute'] = df['time'].dt.minute
```

```python
market_session = []

for hour in df['hour']:

    if hour < 9:

        market_session.append('Pre-Market')

    elif hour < 16:

        market_session.append('Market Hours')

    else:

        market_session.append('After Market')

df['market_session'] = market_session
```

```python
crypto_list = ['BTCUSD', 'ETHUSD']

is_crypto = []

for ticker in df['ticker']:

    if ticker in crypto_list:

        is_crypto.append(True)

    else:

        is_crypto.append(False)

df['is_crypto'] = is_crypto
```

```python
clean_headlines = []

for text in df['headline']:

    cleaned = re.sub(

        r'[^a-zA-Z\s]',

        '',

        text
    )

    cleaned = cleaned.lower()

    cleaned = cleaned.strip()

    clean_headlines.append(cleaned)

df['clean_headline'] = clean_headlines
```

```python
nltk.download('vader_lexicon')
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
```

```python
sia = SentimentIntensityAnalyzer()
```

```python
vader_scores = []
```

```python
for text in df['clean_headline']:

    score = sia.polarity_scores(text)

    compound_score = score['compound']

    vader_scores.append(compound_score)
```

```python
df['vader_score'] = vader_scores
```

```python
df[[

    'headline',

    'vader_score'

]].head()
```

```python
sentiment_labels = []
```

```python
for score in df['vader_score']:

    if score >= 0.50:

        sentiment_labels.append('Very Positive')

    elif score >= 0.05:

        sentiment_labels.append('Positive')

    elif score > -0.05:

        sentiment_labels.append('Neutral')

    elif score > -0.50:

        sentiment_labels.append('Negative')

    else:

        sentiment_labels.append('Very Negative')
```

```python
df['vader_sentiment'] = sentiment_labels
```

```python
from transformers import pipeline
```

```python
bert_classifier = pipeline(

    "sentiment-analysis"
)
```

```python
bert_sentiments = []

bert_scores = []

bert_sentiment_5class = []
```

```python
for text in df['clean_headline']:

    result = bert_classifier(text)

    label = result[0]['label']

    score = result[0]['score']

    bert_sentiments.append(label)

    bert_scores.append(score)

    # Positive
    if label == 'POSITIVE':

        if score >= 0.95:

            bert_sentiment_5class.append('Very Positive')

        elif score >= 0.75:

            bert_sentiment_5class.append('Positive')

        else:

            bert_sentiment_5class.append('Neutral')

    # Negative
    else:

        if score >= 0.95:

            bert_sentiment_5class.append('Very Negative')

        elif score >= 0.75:

            bert_sentiment_5class.append('Negative')

        else:

            bert_sentiment_5class.append('Neutral')
```

```python
df['bert_sentiment'] = bert_sentiments

df['bert_score'] = bert_scores

df['bert_sentiment_5class'] = bert_sentiment_5class
```
